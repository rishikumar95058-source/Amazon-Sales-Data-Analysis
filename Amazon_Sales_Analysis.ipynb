```python
category_summary = (
    df_clean
    .groupby("main_category")
    .agg(
        product_count=("product_id", "count"),
        avg_actual_price=("actual_price", "mean"),
        avg_discounted_price=("discounted_price", "mean"),
        avg_discount=("discount_percentage", "mean"),
        avg_rating=("rating", "mean"),
        total_reviews=("rating_count", "sum")
    )
    .sort_values("product_count", ascending=False)
)

category_summary
```

```python
category_summary = category_summary.round({
    "avg_actual_price": 2,
    "avg_discounted_price": 2,
    "avg_discount": 2,
    "avg_rating": 2,
    "total_reviews": 0
})

category_summary
```

```python
plt.figure(figsize=(10, 6))

category_summary["product_count"].sort_values().plot(kind="barh")

plt.title("Number of Products by Category")
plt.xlabel("Number of Products")
plt.ylabel("Category")

plt.tight_layout()
plt.savefig(
    "images/8_Number_of_Products_by_Category.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

category_summary["avg_discounted_price"].sort_values().plot(kind="barh")

plt.title("Average Discounted Price by Category")
plt.xlabel("Average Discounted Price (₹)")
plt.ylabel("Category")

plt.tight_layout()
plt.savefig(
    "images/9_Average_Discounted_Price_by_Category.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

category_summary["avg_rating"].sort_values().plot(kind="barh")

plt.title("Average Product Rating by Category")
plt.xlabel("Average Rating")
plt.ylabel("Category")

plt.tight_layout()
plt.savefig(
    "images/10_Average_Product_Rating_by_Category.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
